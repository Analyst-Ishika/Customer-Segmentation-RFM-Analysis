# 📊 Step 3: Building the RFM Table
---

```python
snapshot_date = df["invoice_date"].max().normalize() + pd.Timedelta(days=1)
recency = df[df["is_return"] == False].groupby("customer_id")["invoice_date"].max().reset_index()
recency["recency"] = (snapshot_date - recency["invoice_date"].dt.normalize()).dt.days
recency.drop(["invoice_date"], axis=1, inplace=True)
recency.head()
```

```python
frequency = df[df["is_return"] == False].groupby("customer_id")["invoice"].nunique().reset_index()
frequency.columns = ["customer_id","frequency"]
frequency.head()
```

```python
monetary = df.groupby("customer_id")["amount"].sum().reset_index()
monetary.columns = ["customer_id","monetary"]
monetary.head()
```

```python
rfm = recency.merge(frequency, on="customer_id").merge(monetary, on="customer_id")
rfm.head()
```

```python
rfm.shape
```

```python
rfm.describe()
```

```python
rfm.info()
```

**Build RFM scores**

```python
rfm["r_score"] = pd.qcut(rfm["recency"], 5, labels=[5, 4, 3, 2, 1]).astype(int)
rfm["f_score"] = pd.cut(rfm["frequency"],bins=[0, 1, 2, 5, 12, float("inf")],labels=[1, 2, 3, 4, 5]).astype(int)
rfm["m_score"] = pd.qcut(rfm["monetary"], 5, labels=[1, 2, 3, 4, 5]).astype(int)
rfm["rfm_score"] = rfm["r_score"].astype(str) + rfm["f_score"].astype(str) + rfm["m_score"].astype(str)
rfm["rfm_sum"] = rfm["r_score"] + rfm["f_score"] + rfm["m_score"]
rfm.head()
```

**Why qcut for Recency and Monetary, and cut for Frequency**

From `rfm.describe()` I saw that the three metrics are not evenly spread:
- **Monetary** is very skewed: the median is about £818, but the mean is about £2,572 and the maximum is far above the 75th percentile. A few very large customers pull the average up.
- **Recency** ranges from 1 to 739 days, spread across the whole period.
- **Frequency** has a 25th percentile of 1, so a large group of customers bought only once.

**Recency and Monetary → `pd.qcut`.** Because of the skew, equal-width bins would put almost every customer in the first bucket. `qcut` splits customers by rank into 5 equal groups, so each score has about the same number of customers. For Recency I reversed the labels (5 = most recent), because fewer days since the last purchase is better.

**Frequency → `pd.cut` with my own bins.** So many customers share the same low values (1, 2, 3 purchases) that equal-sized groups are not possible without splitting customers who behave identically. I chose my own bins (1 | 2 | 3-5 | 6-12 | 13+) so that customers with the same number of purchases always get the same score. The last edge is `inf` so that no customer is left without a score.

## **Segmentation**

```python
grid = pd.crosstab(rfm["r_score"], rfm["f_score"]).sort_index(ascending=False)
grid
```

```python
conditions = [
    (rfm["r_score"] >= 4) & (rfm["f_score"] == 5),   # Champions
    (rfm["r_score"] <= 2) & (rfm["f_score"] >= 3),   # At-Risk
    (rfm["r_score"] <= 2) & (rfm["f_score"] <= 2),   # Lost
    (rfm["r_score"] >= 4) & (rfm["f_score"] == 1),   # New
    (rfm["r_score"] >= 3) & (rfm["f_score"] >= 3),   # Loyal
]
names = ["Champions","At-Risk","Lost","New","Loyal"]
rfm["segment"] = np.select(conditions,names,default="Others")
rfm["value_tier"] = np.where(rfm["m_score"] >= 4, "High","Low")
rfm.head()
```

```python
rfm_analysis = rfm.groupby("segment").agg(
    customers = ("customer_id", "count"),
    avg_recency = ("recency", "mean"),
    avg_frequency = ("frequency", "mean"),
    avg_monetary = ("monetary", "mean"),
    median_monetary = ("monetary", "median"),
    total_revenue = ("monetary", "sum"),
).round(1).reset_index()
rfm_analysis["customer_%"] = (rfm_analysis["customers"] / rfm_analysis["customers"].sum() * 100).round(1)
rfm_analysis["total_revenue_%"] = (rfm_analysis["total_revenue"] / rfm_analysis["total_revenue"].sum() * 100).round(1)
rfm_analysis
```

```python
rfm_analysis["value_index"] = (rfm_analysis["total_revenue_%"] / rfm_analysis["customer_%"]).round(1)
rfm_analysis["avg_order_value"] = (rfm_analysis["avg_monetary"] / rfm_analysis["avg_frequency"]).round(1)
rfm_analysis
```

# 💰 Step 4: Campaign ROI Model & Budget Allocation
---

```python
assumptions = {
    "margin":0.30,
    "cost_per_contact":3.0,
    "extra_response":{
        "Champions":0.03, "Loyal":0.06, "At-Risk":0.08, "Lost":0.01, "New":0.05, "Others":0.04}
}

roi = rfm_analysis[["segment", "customers", "avg_frequency", "avg_monetary", "avg_order_value"]].copy()
roi["extra_response"] = roi["segment"].map(assumptions["extra_response"])
roi["profit_per_cust"] = roi["avg_order_value"] * roi["extra_response"] * assumptions["margin"]
roi["break_even_cost"] = roi["profit_per_cust"]
roi["roi_%"] = (roi["profit_per_cust"] - assumptions["cost_per_contact"]) / assumptions["cost_per_contact"] * 100
roi["net_profit"] = roi["customers"] * (roi["profit_per_cust"] - assumptions["cost_per_contact"])
roi.round(2).sort_values("roi_%", ascending=False)
```

```python
price = {"email":0.03, "sms":0.05, "ads":0.75, "mail":1.00}   # per customer

assum_2 = {
    "margin":0.30,
    "extra_response": {
        "Champions":0.03, "Loyal":0.06, "At-Risk":0.08, "Lost":0.01, "New":0.05, "Others":0.04},
    "channel_mix": {
        "Champions": ["email", "sms", "mail"],
        "Loyal": ["email", "sms"],
        "At-Risk": ["email", "sms", "ads"], 
        "Lost": ["email"],
        "New": ["email", "sms", "ads"],
        "Others": ["email", "ads"]},
    "incentive_cost": {
        "Champions":1, "Loyal":1, "At-Risk":1, "Lost":0, "New":1, "Others":1}
}

roi_2 = rfm_analysis[["segment", "customers", "avg_frequency", "avg_monetary", "avg_order_value"]].copy()

roi_2["extra_response"] = roi_2["segment"].map(assum_2["extra_response"])
roi_2["profit_per_cust"] = roi_2["avg_order_value"] * roi_2["extra_response"] * assum_2["margin"]
roi_2["channel_cost"] = roi_2["segment"].map(lambda s: sum(price[c] for c in assum_2["channel_mix"][s]))
roi_2["incentive"] = roi_2["segment"].map(assum_2["incentive_cost"])
roi_2["cost_per_cust"] = roi_2["channel_cost"] + roi_2["incentive"]
roi_2["max_channel_cost"] = (roi_2["profit_per_cust"] - roi_2["incentive"])/2
assert (roi_2["channel_cost"] <= roi_2["max_channel_cost"]).all(), "The cost for a segment is above the ceiling!"

roi_2["roi_%"] = (roi_2["profit_per_cust"] - roi_2["cost_per_cust"]) / roi_2["cost_per_cust"] * 100
roi_2["total_spend"] = roi_2["cost_per_cust"] * roi_2["customers"]
roi_2["net_profit"] = roi_2["customers"] * (roi_2["profit_per_cust"] - roi_2["cost_per_cust"])

roi_2["spend_%"] = roi_2["total_spend"] / (roi_2["total_spend"].sum()) * 100
roi_2["net_profit_%"] = roi_2["net_profit"] / (roi_2["net_profit"].sum()) * 100
roi_2.round(2)
```

```python
final = roi_2[["segment", "customers", "profit_per_cust", "cost_per_cust", "roi_%", "total_spend", "net_profit", "spend_%", "net_profit_%"]]
final.round(2).sort_values("net_profit", ascending=False)
```

## Backtest: Validating the Response-Rate Assumption

```python
df["invoice_date"].agg(["min","max"])
```

```python
snapshot = pd.Timestamp("2010-12-09")

before = df[df["invoice_date"] <  snapshot]
after  = df[df["invoice_date"] >= snapshot]

cust_before = set(before["customer_id"])
cust_after  = set(after["customer_id"])

print("Before Snapshot:", len(cust_before))
print("Returned customers:", len(cust_before & cust_after))
print("Returned customers %:", round((len(cust_before & cust_after) / len(cust_before) * 100)))
```

**Build RFM for before and after customers**

```python
def build_rfm(df, snapshot_date):

    # --- Recency ---
    recency = df[df["is_return"] == False].groupby("customer_id")["invoice_date"].max().reset_index()
    recency["recency"] = (snapshot_date - recency["invoice_date"].dt.normalize()).dt.days
    recency.drop(["invoice_date"], axis=1, inplace=True)

    # --- Frequency ---
    frequency = df[df["is_return"] == False].groupby("customer_id")["invoice"].nunique().reset_index()
    frequency.columns = ["customer_id","frequency"]

    # --- Monetary ---
    monetary = df.groupby("customer_id")["amount"].sum().reset_index()
    monetary.columns = ["customer_id","monetary"]

    # --- RFM ---
    rfm = recency.merge(frequency, on="customer_id").merge(monetary, on="customer_id")
    
    rfm["r_score"] = pd.qcut(rfm["recency"], 5, labels=[5, 4, 3, 2, 1]).astype(int)
    rfm["f_score"] = pd.cut(rfm["frequency"],bins=[0, 1, 2, 5, 12, float("inf")],labels=[1, 2, 3, 4, 5]).astype(int)
    rfm["m_score"] = pd.qcut(rfm["monetary"], 5, labels=[1, 2, 3, 4, 5]).astype(int)
    rfm["rfm_score"] = rfm["r_score"].astype(str) + rfm["f_score"].astype(str) + rfm["m_score"].astype(str)
    rfm["rfm_sum"] = rfm["r_score"] + rfm["f_score"] + rfm["m_score"]

    # --- Segmentation ---
    conditions = [
        (rfm["r_score"] >= 4) & (rfm["f_score"] == 5),   # Champions
        (rfm["r_score"] <= 2) & (rfm["f_score"] >= 3),   # At-Risk
        (rfm["r_score"] <= 2) & (rfm["f_score"] <= 2),   # Lost
        (rfm["r_score"] >= 4) & (rfm["f_score"] == 1),   # New
        (rfm["r_score"] >= 3) & (rfm["f_score"] >= 3),   # Loyal
    ]
    names = ["Champions","At-Risk","Lost","New","Loyal"]
    rfm["segment"] = np.select(conditions,names,default="Others")
    return rfm
```

```python
before = df[df["invoice_date"] <  snapshot]
after  = df[(df["invoice_date"] >= snapshot) & (df["is_return"] == False)]

rfm_past = build_rfm(before, snapshot)

returned_ids = set(after["customer_id"])
rfm_past["returned"] = rfm_past["customer_id"].isin(returned_ids)

rfm_past.groupby("segment")["returned"].agg(customers = "count", returned = "sum", return_rate = "mean").round(2).sort_values("return_rate", ascending=False)
```

```python
baseline = rfm_past.groupby("segment")["returned"].mean().rename("baseline").reset_index()
baseline
```

```python
t = roi_2[["segment", "customers", "avg_order_value", "cost_per_cust"]].merge(baseline, on="segment")
t["headroom"] = 1 - t["baseline"]
t
```

```python
rows = []
for x in [0.05, 0.10, 0.15]:
    s = t.copy()                                  
    s["lift"] = s["headroom"] * x
    s["profit_per_cust"] = s["avg_order_value"] * s["lift"] * assum_2["margin"]
    s["roi_%"] = (s["profit_per_cust"] - s["cost_per_cust"]) / s["cost_per_cust"] * 100
    s["net_profit"] = (s["customers"] * (s["profit_per_cust"] - s["cost_per_cust"])).round(1)
    s["X"] = x
    rows.append(s[["X", "segment", "lift", "roi_%", "net_profit"]])

pd.concat(rows).pivot(index="segment", columns="X", values=["roi_%", "net_profit"]).round(1)
```

```python
t["implied_conversion"] =  ((roi_2["extra_response"] / t["headroom"]) * 100).round(1)
t.sort_values("implied_conversion")
```

```python
x = 0.10
base = t.copy()
base["lift"] = base["headroom"] * x
base["profit_per_cust"] = base["avg_order_value"] * assum_2["margin"] * base["lift"]
base["roi_%"] = (base["profit_per_cust"] - base["cost_per_cust"]) / base["cost_per_cust"] * 100
base["total_spend"] = base["cost_per_cust"] * base["customers"]
base["net_profit"] = base["customers"] * (base["profit_per_cust"] - base["cost_per_cust"])
base[["segment", "roi_%", "total_spend", "net_profit"]].round(1).sort_values("roi_%", ascending=False)
```

## Budget Allocation

```python
budget = 3000
alloc = base.sort_values("roi_%", ascending=False).copy()
alloc["spent_before"] = alloc["total_spend"].cumsum().shift(fill_value=0)
alloc["funded_spend"] = np.minimum(np.maximum(budget - alloc["spent_before"],0), alloc["total_spend"])
alloc["funded_%"] = (alloc["funded_spend"] / alloc["total_spend"]) * 100
alloc["funded_cust"] = alloc["customers"] * alloc["funded_%"] / 100
alloc["funded_net_profit"] = alloc["net_profit"] * alloc["funded_%"] / 100
```

```python
alloc = alloc[["segment", "roi_%", "funded_spend", "funded_%", "funded_cust", "funded_net_profit"]].round(1)
alloc
```

```python
headroom_by_r = 1 - rfm_past[rfm_past["segment"] == "Loyal"].groupby("r_score")["returned"].mean()

l = rfm[rfm["segment"] == "Loyal"].copy()
l["headroom"] = l["r_score"].map(headroom_by_r)
l["avg_order"] = l["monetary"] / l["frequency"]
l["exp_profit"] = l["headroom"] * 0.10 * l["avg_order"] * 0.30

target_l = l.sort_values("exp_profit",ascending=False).head(266)
```

## Key Findings

**1. Revenue is concentrated in a small group.**
Champions are 9.5% of customers but generate 53% of revenue. Loyal customers add another 30.6%. Together, 44% of customers produce 84% of revenue.

**2. Retention ROI depends on headroom, not on how valuable a customer is.**
Champions are the most valuable segment, but 97% of them returned in the backtest even without a campaign, so a retention campaign can add very little. At a 10% campaign success rate their ROI is -77%, and it stays negative at 5% and 15% too.

**3. Recommended allocation of a £3,000 budget (base case, X = 10%):**

**4. Expected net profit is about £4,200, and up to £13,000 if Lost responds as assumed.**
About two-thirds of the £13,000 comes from Lost, which costs only £45 and rests on one assumption (a 6% lift from a £0.03 email). The £4,200 figure excludes it and is the more conservative number to plan around.

**5. At-Risk and Loyal are the most sensitive to the success-rate assumption.**
At 5% both lose money (-6% and -29%). They pay off from roughly 10%. New and Others are positive even at 5%.

**Recommendation:** Fund At-Risk, Others and New in full, fund only the best Loyal customers with the remaining budget, and spend nothing on Champions in this campaign. Champions need a different lever (spend per order, e.g. VIP perks), not retention. Before scaling, run an A/B test with a holdout group to measure the real success rate.

## Limitations

- **The success rate (X = 10%) is an assumption.** The dataset contains no campaign, so it cannot be measured here. It should be validated with an A/B test (campaign group vs. holdout group).
- **Lost drives most of the headline profit.** Its ROI rests on a £0.03 email producing a 6% lift. In the backtest 37% of "Lost" customers came back on their own, so the label is also imperfect.
- **Allocation is done at segment level.** Ranking individual customers by expected profit (headroom × order value) would probably shift the budget. For example, the selected 266 Loyal customers return ~245% ROI, higher than At-Risk (88%).
- **Headroom is estimated from one backtest window** (snapshot 9 Dec 2010) and by recency score, not per customer.
- **UK customers only** (about 92% of rows). Customers without an ID (about 24% of rows) were excluded, so the results do not cover them.
